O(n) example: Only one for loop O(n)

```python
def print_items(n):
    for i in range(n):
        print(i)


print_items(10)
```

O(n^2) one for loop inside the other loop, n*n=n^2 =O(n^2)

```python
def print_items(n):
    for i in range(n):
        for j in range(n):
            print(i,j)

print(print_items(10))
```
